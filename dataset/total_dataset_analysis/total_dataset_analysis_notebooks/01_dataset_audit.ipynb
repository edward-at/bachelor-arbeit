Import libraries and define folders

```python
from pathlib import Path

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Detect project folder
PROJECT_DIR = Path.cwd()

# If the notebook runs from the notebooks folder, move one level up
if PROJECT_DIR.name == "notebooks":
    PROJECT_DIR = PROJECT_DIR.parent

DATA_RAW = PROJECT_DIR / "data_raw"
DATA_PROCESSED = PROJECT_DIR / "data_processed"
FIGURES = PROJECT_DIR / "figures"
TABLES = PROJECT_DIR / "tables"

# Create output folders if they do not exist
DATA_PROCESSED.mkdir(exist_ok=True)
FIGURES.mkdir(exist_ok=True)
TABLES.mkdir(exist_ok=True)

print("Project directory:", PROJECT_DIR)
print("Raw data folder:", DATA_RAW)
print("Processed data folder:", DATA_PROCESSED)
```

Find Excel files in data_raw

```python
excel_files = sorted(DATA_RAW.glob("*.xlsx"))

print("Excel files found:")
for i, file in enumerate(excel_files, start=1):
    print(f"{i}. {file.name}")
```

Inspect sheet names

```python
for file in excel_files:
    print("\n" + "=" * 80)
    print(f"File: {file.name}")
    print("=" * 80)
    
    xls = pd.ExcelFile(file)
    
    for sheet in xls.sheet_names:
        print("-", sheet)
```

Preview the raw structure of each sheet


```python
for file in excel_files:
    xls = pd.ExcelFile(file)
    
    for sheet in xls.sheet_names:
        print("\n" + "=" * 100)
        print(f"File: {file.name}")
        print(f"Sheet: {sheet}")
        print("=" * 100)
        
        preview = pd.read_excel(file, sheet_name=sheet, header=None, nrows=10)
        display(preview)
```
